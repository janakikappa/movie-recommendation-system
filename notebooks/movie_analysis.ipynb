# Movie Recommendation System - Analysis Notebook
Run from the project root after placing the TMDB CSV files in `data/raw/`. Start Jupyter from the project root.

```python
import sys, os
os.chdir('..') if os.path.basename(os.getcwd())=='notebooks' else None
sys.path.insert(0, os.getcwd())
from src.data_preprocessing import load_raw_data, understand_data, clean_data, save_clean
from src.eda import run_eda
from src.feature_engineering import add_features, save_features
from src.recommendation_model import build_model, save_model, Recommender, evaluate
```

## 1. Data collection & understanding

```python
movies, credits, merged = load_raw_data()
understand_data(merged)
```

## 2. Cleaning
Duplicates removed, text gaps filled, JSON columns parsed (see comments in `src/data_preprocessing.py`).

```python
clean = clean_data(merged)
save_clean(clean)
clean.head()
```

## 3. EDA
Charts are saved to `assets/`.

```python
run_eda(clean)
from IPython.display import Image, display
for f in ['top_popular','rating_distribution','genre_counts']:
    display(Image(f'assets/{f}.png'))
```

## 4. Feature engineering

```python
data = add_features(clean)
save_features(data)
data[['title','combined_features']].head()
```

## 5. TF-IDF + cosine similarity

```python
vectorizer, similarity = build_model(data)
save_model(vectorizer, similarity)
similarity.shape
```

## 6. Recommend & evaluate

```python
rec = Recommender(data, similarity)
rec.recommend('Avatar', 5)[['title','vote_average','similarity']]
```
